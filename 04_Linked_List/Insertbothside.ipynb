```python
# Add the new elements in both the end
# Step 1: Create a Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Step 2: Create the existing nodes
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)

# Step 3: Connect the existing nodes
node1.next = node2
node2.next = node3
node3.next = node4

head = node1

# Step 4: Add three elements at the beginning
for value in [50, 60, 70]:
    new_node = Node(value)
    new_node.next = head
    head = new_node

# Step 5: Add three elements at the end
for value in [80, 90, 100]:
    new_node = Node(value)
    current = head

    while current.next is not None:
        current = current.next

    current.next = new_node

# Step 6: Display the linked list
current = head

while current is not None:
    print(current.data, end=" -> ")
    current = current.next

print("None")
```

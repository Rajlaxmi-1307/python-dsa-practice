```python
# Insertion at end
# Step 1: Create a Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Step 2: Create nodes
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
node4 = Node(40)


# Step 3: Connect nodes
node1.next = node2
node2.next = node3
node3.next = node4


# Step 4: Insert a new node at the end
head = node1

new_node = Node(50)
current = head

while current.next is not None:
    current = current.next

current.next = new_node


# Step 5: Display the linked list
current = head

while current is not None:
    print(current.data, end=" -> ")
    current = current.next

print("None")
```

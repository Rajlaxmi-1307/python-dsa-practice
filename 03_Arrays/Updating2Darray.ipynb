```python
#Updating elements 2D
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("Original 2D Array:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()

matrix[1][1] = 55

print("\n Updated 2D Array:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

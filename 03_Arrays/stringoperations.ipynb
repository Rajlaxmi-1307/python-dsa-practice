```python
S = "HELLO"

print("1. String Creation:")
print(S)

print("\n2. Accessing Characters:")
print(S[0])
print(S[1])
print(S[-1])

print("\n3. Traversing String:")
for ch in S:
    print(ch)

print("\n4. Finding Length:")
print(len(S))

print("\n5. Searching Character:")
if 'L' in S:
    print("Character found")
else:
    print("Character not found")

print("\n6. Counting Character:")
print(S.count('L'))

print("\n7. Reversing String:")
print(S[::-1])

print("\n8. String Concatenation:")
S2 = " WORLD"
print(S + S2)

print("\n9. Comparing Strings:")
if S == "HELLO":
    print("Strings are equal")
else:
    print("Strings are not equal")

print("\n10. Changing Case:")
print(S.upper())
print(S.lower())

print("\n11. Checking Palindrome:")
if S == S[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

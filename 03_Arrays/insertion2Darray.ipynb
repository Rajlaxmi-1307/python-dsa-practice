```python
#Insertion in a 2D Array
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("Original 2D Array:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()

matrix.insert(2, [100,110,120])

print("\n After Insertion:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

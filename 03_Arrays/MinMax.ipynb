```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

maximum = matrix[0][0]
minimum = matrix[0][0]

for row in matrix:
    for value in row:

        if value > maximum:
            maximum = value

        if value < minimum:
            minimum = value

print("Maximum:", maximum)
print("Minimum:", minimum)
```

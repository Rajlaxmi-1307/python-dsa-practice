```python
#Deletion in a  2D Array
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("Original 2D Array:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()

matrix.pop(1)

print("\n After Deletion:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()
```

```python
# Import NumPy library
import numpy as np

# Create a NumPy array containing positive, negative, and zero values
arr = np.array([12, -3, 7, 0, 15, -3, 15, 8, 10])

# Print the original array
print("Original Array:", arr)

# Traverse the array and print each element
for x in arr:
    print(x)

# Calculate and print the sum of all elements
total_sum = np.sum(arr)
print("SUM:", total_sum)

# Calculate and print the average of all elements
average = np.mean(arr)
print("Average:", average)

# Find and print the minimum and maximum values
min_value = np.min(arr)
max_value = np.max(arr)

print("Min:", min_value)
print("Max:", max_value)

# Filter and print all even numbers
even = arr[arr % 2 == 0]
print("Even no.:", even)

# Filter and print all odd numbers
odd = arr[arr % 2 != 0]
print("Odd no.:", odd)

# Count the number of even and odd elements
count_even = np.sum(arr % 2 == 0)
count_odd = np.sum(arr % 2 != 0)

print("Count_even:", count_even)
print("Count_odd:", count_odd)

# Count the total number of elements in the array
count_all = arr.size
print("Count all:", count_all)

# Count positive, negative, and zero elements
positive = np.sum(arr > 0)
print("Positive:", positive)

negative = np.sum(arr < 0)
print("Negative:", negative)

zero = np.sum(arr == 0)
print("Zero:", zero)

# Reverse the array using slicing
reverse = arr[::-1]
print("Reverse:", reverse)

# Get unique elements and sort them in ascending order
sort = np.sort(np.unique(arr))

# Find the second smallest and second largest unique values
second_smallest = sort[1]
second_largest = sort[-2]

print("Second smallest:", second_smallest)
print("Second largest:", second_largest)

# Extract and print unique elements (remove duplicates)
duplicates = np.unique(arr)
print("Unique elements:", duplicates)
```

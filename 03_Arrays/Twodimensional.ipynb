```python
#Create a 2D Array named matrix
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

print("2D Array :")
print(matrix)

print("First Elements:")
print(matrix[0][0])

print("Middle Elements:")
print(matrix[1][1])

print("Last Elements:")
print(matrix[2][2])

print("Only Row 2:")
print(matrix[2])

print("One Column:")
for row in matrix:
    print(row[1])
```

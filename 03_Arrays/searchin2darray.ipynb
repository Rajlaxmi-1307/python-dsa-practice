```python
#Searching in a 2D Array
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

search = 50

found = False

for row in matrix:
    for value in row:
        if value == search:
            found = True

if found:
    print(search, "is found")
else:
    print(search, "is not found")
```

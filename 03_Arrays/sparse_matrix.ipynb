```python
# Sparse Matrix
matrix = [
    [1, 0, 0, 0],
    [0, 2, 0, 0],
    [0, 0, 3, 0],
    [0, 0, 0, 4],
    [5, 0, 0, 0]
]

print("Original Matrix:")

for row in matrix:
    for value in row:
        print(value, end=" ")
    print()

sparseMatrix = []

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if matrix[i][j] != 0:

            temp = []

            temp.append(i)
            temp.append(j)
            temp.append(matrix[i][j])

            sparseMatrix.append(temp)

print("\nSparse Matrix:")

for row in sparseMatrix:
    for value in row:
        print(value, end=" ")
    print()
```
